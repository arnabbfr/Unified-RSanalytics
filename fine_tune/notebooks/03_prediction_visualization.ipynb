# 03. Qualitative Prediction Visualizer & Error Analysis

This notebook displays standardized 4-panel visual comparison cards:
- **SAR Input** (False Color: Red=VV, Green=VH, Blue=VV/VH Ratio)
- **Ground Truth Mask** (Binary Flood Boundary)
- **Foundation Model Predicted Mask**
- **Confusion Overlay** (Green=True Positive, Red=False Alarm, Blue=Missed Flood)

```python
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image

PROJECT_ROOT = Path("..").resolve()
RESULTS_DIR = PROJECT_ROOT / "fine_tune" / "results"

models = ["terramind", "prithvi", "satmaepp", "gfm"]

for m in models:
    pred_dir = RESULTS_DIR / m / "predictions"
    if pred_dir.is_dir():
        cards = sorted(list(pred_dir.glob("*_comparison.png")))
        if cards:
            print(f"\n{'='*70}\nPREDICTIONS FOR MODEL: {m.upper()} ({len(cards)} samples)\n{'='*70}")
            for c_path in cards[:4]:
                img = Image.open(c_path)
                plt.figure(figsize=(18, 5))
                plt.imshow(img)
                plt.axis("off")
                plt.title(f"{m.upper()} — {c_path.name}", fontsize=12)
                plt.show()
```
